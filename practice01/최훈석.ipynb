```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
sc = spark.sparkContext
```

```python
rdd = sc.textFile("sherlock.txt")
```

```python
rdd.flatMap(lambda line:line.split(" "))\
.filter(lambda word:len(word) >=2)\
.map(lambda word:word.lower())\
.map(lambda word:(word[:2],1))\
.reduceByKey(lambda a,b:a+b) \
.sortBy(lambda x:x[1],ascending=True) \
.filter(lambda x:x[1] < 32) \
.map(lambda x:x[1])\
.sum()
```

```python
rdd = sc.textFile("int-pair-input.txt")

rdd2 = rdd.flatMap(lambda line:line.split("*,*"))\
.map(lambda a:a.strip("()").split(", ")) \
.map(lambda x:(int(x[0]),int(x[1])))\

first_value = rdd2.map(lambda x:x[0])

first_element = first_value.map(lambda x:(x,1))\
    .reduceByKey(lambda a,b:a+b)

print(
    first_value.first(),
    first_value.max(),
    first_value.min(),
    first_element.filter(lambda x:x[0] == 1).map(lambda x:x[1]).first(),
    first_element.filter(lambda x:x[0] == 2).map(lambda x:x[1]).first()
)
print(
    first_value.first() + \
    first_value.max() + \
    first_value.min() + \
    first_element.filter(lambda x:x[0] == 1).map(lambda x:x[1]).first() + \
    first_element.filter(lambda x:x[0] == 2).map(lambda x:x[1]).first()
)

# rdd.flatMap(lambda line:line.split("(*,*)"))\
# .take(5)
```

#Q3

```python
years =[2015,2016,2017,2018,2019]
data={}

for year in years:
  data[year] = sc.textFile(f"{year}.csv")
  data[year] = data[year].map(lambda line:line.split(","))\
  .filter(lambda x:x[0] != "Country")\
  .map(lambda x:(x[0],float(x[2])))
  # print(data[year].take(10))

data = data[2015].union(data[2016]).union(data[2017]).union(data[2018]).union(data[2019])
data = data.map(lambda x: (x[0],[x[1]]))\
.reduceByKey(lambda a,b:a+b)\
.filter(lambda x:len(x[1]) == 5 and x[1][0]<=x[1][1]<=x[1][2]<=x[1][3]<=x[1][4])\
.map(lambda x:x[0])

print(data.collect())
# data = data.filter(lambda x: x[0])
```
