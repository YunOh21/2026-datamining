```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .master("local[*]") \
    .getOrCreate()

sc = spark.sparkContext
```

```python
rdd = sc.textFile("data01/int-pair-input.txt")

# sol
pairs = rdd.map(lambda line: line.replace("(", "").replace(")", "").split(",")) \
           .map(lambda x: (int(x[0]), int(x[1])))

# 첫 번째 정수 쌍의 첫 번째 원소
a = pairs.first()[0]

# max, min, count1, count2
result = pairs.map(lambda x: (
        x[0],                       # max 후보
        x[0],                       # min 후보
        1 if x[0] == 1 else 0,     # count1
        1 if x[0] == 2 else 0      # count2
    )) \
    .reduce(lambda x, y: (
        x[0] if x[0] > y[0] else y[0],
        x[1] if x[1] < y[1] else y[1],
        x[2] + y[2],
        x[3] + y[3]
    ))

b = result[0]
c = result[1]
d = result[2]
e = result[3]

print(a, b, c, d, e)
print(a + b + c + d + e)
```
