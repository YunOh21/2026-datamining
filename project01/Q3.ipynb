```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .master("local[*]") \
    .getOrCreate()

sc = spark.sparkContext
```

```python
years = [2015, 2016, 2017, 2018, 2019]
data = {}

for year in years:
    data[year] = sc.textFile(f"data01/{year}.csv")
    data[year] = data[year].map(lambda line: line.split(","))\
        .filter(lambda x: x[0] != "Country")\
        .map(lambda x: (x[0], (year, float(x[2]))))

data = data[2015].union(data[2016]).union(data[2017]).union(data[2018]).union(data[2019])

data = data.map(lambda x: (x[0], [x[1]]))\
    .reduceByKey(lambda a, b: a + b)\
    .mapValues(lambda x: sorted(x, key=lambda t: t[0]))\
    .filter(lambda x: len(x[1]) == 5 and
                      x[1][0][1] <= x[1][1][1] <= x[1][2][1] <= x[1][3][1] <= x[1][4][1])\
    .map(lambda x: x[0])

print(data.collect())
```
